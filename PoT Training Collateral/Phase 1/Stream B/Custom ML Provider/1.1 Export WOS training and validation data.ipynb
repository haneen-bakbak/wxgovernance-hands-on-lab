```python
import pandas as pd
import numpy as np
```

```python
df= pd.read_csv("AMZN_Stock_Updated_V2.csv")
df.set_index('Date',inplace = True)# Set the date to be the index
df.index =  pd.to_datetime(df.index,format='%Y-%m-%d')
df.head()
```

```python
train_Ama = df['High'].iloc[:-4]

X_train=[]
y_train=[]

for i in range(2, len(train_Ama)):
    X_train.append(train_Ama[i-2:i])
    y_train.append(train_Ama[i])

X_train, y_train = np.array(X_train), np.array(y_train)
X_train = np.reshape(X_train, (X_train.shape[0], X_train.shape[1], 1))

train_df= pd.DataFrame(X_train.squeeze(), columns= ["lag_2", "lag_1"])
train_df.insert(loc=2, column="y", value=y_train)
train_df
```

```python
import math
train_len = math.ceil(len(train_Ama)*0.8)
train_len

X_val=[]
# Y_val=[] 
test_data = train_Ama[train_len-2:]
X_val=[]
Y_val=[] 

for i in range(2, len(test_data)):
    X_val.append(test_data[i-2:i])
    Y_val.append(test_data[i])

X_val, Y_val = np.array(X_val), np.array(Y_val)
X_val = np.reshape(X_val, (X_val.shape[0], X_val.shape[1], 1))

val_df= pd.DataFrame(X_val.squeeze(), columns= ["lag_2", "lag_1"])
val_df.insert(loc=2, column="y", value=Y_val)
val_df
```

### Save Data

```python
train_df.to_csv("train.csv", index=False)
val_df.to_csv("val.csv", index=False)
```
